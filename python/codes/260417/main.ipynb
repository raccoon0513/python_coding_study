```python
## 제품번호 1~10

def count_gloves(glove_list):
    count = [0]*11
    for i in range(len(glove_list)):
        count[glove_list[i]] +=1
    return count

def count_pair(left_count, right_count):
    count = 0
    for i in range(len(left_count)):
        count+= min(left_count[i], right_count[i])
    return count
left_gloves = [2,1,2,2,4]
right_gloves = [1,2,2,4,4,7]

left_count = count_gloves(left_gloves)
right_count = count_gloves(right_gloves)
count = count_pair(left_count, right_count)

print(count)
```

```python
count_3, count_5 = 0,0
arr = [2,3,6,9,12,15,10,20,22,25]
for i in arr:
    if i%3 ==0 : count_3+=1
    if i%5 ==0 : count_5+=1

if count_3 == count_5:
    print("둘의 수가 같습니다")
else :
    print(f"{'3' if count_3 > count_5 else '5'}의  배수가 더 많습니다")
```

```python
#몇번공격해야하나
attack = 30
recovery = 10
hp = 60
count = 0
while(hp>0):
    hp-=attack
    count+=1
    if hp <=0:
        break
    hp+=recovery
print(count)
```

```python
floors = [1, 2, 5, 4, 2]
count = 0
for i in range(1, len(floors)):
    count += abs(floors[i-1] - floors[i])
print(count)
```

```python
votes =  [2, 5, 3, 4, 1, 5, 1, 5, 5, 3]
votes_len = 10
N = 5
K = 2

def solution(votes, N, K):
    counts = [0] * (N+1)
    K_count = 0
    for i in votes:
        counts[i]+=1
    for i in counts:
        if i == K: K_count+=1

    return K_count

print(solution(votes, N, K))
```
