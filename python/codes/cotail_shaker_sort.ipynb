```python
def sort(arr_original):
    arr = arr_original.copy()
    start_index = 0
    end_index = len(arr)-1
    reverse = 1
    while(start_index < end_index):
        not_changed = True
        if reverse == 1:
            for i in range(start_index, end_index):
                if arr[i] > arr[i+1]:
                    arr[i], arr[i+1], not_changed = arr[i+1], arr[i], False
            end_index-=1
            reverse = -1
            if not_changed:
                return arr
        elif reverse == -1:
            for i in range(start_index, end_index)[::-1]:
                if arr[i] > arr[i+1]:
                    arr[i], arr[i+1], not_changed = arr[i+1], arr[i], False
            start_index+=1
            reverse = 1
            if not_changed:
                return arr

print(sort(arr))
```

```python
def sort(arr_original):
    arr = arr_original.copy()
    reverse = 1 # 1 or -1
    indexes = {
        -1 : 0 # start_index
        ,1 : len(arr) - 1 # end_index
    }
    while indexes[-1] < indexes[1]:
        not_changed = True
        for i in range(indexes[-1*reverse], indexes[1*reverse], reverse):
            if arr[i]*reverse > arr[i+reverse] * reverse:
                arr[i+reverse], arr[i] = arr[i], arr[i+reverse]
                not_changed = False
        indexes[reverse] -= reverse
        reverse *=-1
        if not_changed:
            return arr
    return arr
print(sort(arr))
```

```python

def sort(arr):
    arr_c, r = arr.copy(), 1
    # 배열 길이만큼 돌거나, 정렬이 이미 끝났으면(체크용) 종료
    for _ in range(len(arr)):
        swapped = False
        # r=1이면 정방향, r=-1이면 역방향 range를 한 줄로 생성
        for i in range(0, len(arr)-1)[::r]:
            if (arr_c[i] - arr_c[i+1]) * r > 0:
                arr_c[i], arr_c[i+1], swapped = arr_c[i+1], arr_c[i], True
        # 교환이 없었거나(정렬완료), r을 뒤집어 다음 턴 준비
        if not swapped and r == -1: break
        r *= -1
    return arr_c

# 테스트
print(sort([5, 1, 4, 2, 8, 0, 2]))  # [0, 1, 2, 2, 4, 5, 8]
```
