```python
# 3번
def solution(scores : list, scores_len):
    scores_copy = scores.copy()
    scores_copy.remove(min(scores_copy))
    scores_copy.remove(max(scores_copy))

    return int(sum(scores_copy) / len(scores_copy))

print(solution([35, 28, 98, 34, 20, 50, 85, 74, 71, 7], 10))
print(solution([1,1,1,1,1], 5))
```

```python
# 4번
def solution(words, words_len, word):
    count = 0
    for i in words:
        for j in range(len(i)):
            if i[j]!= word[j] : count+=1
    return count

print(solution(["CODE", "COED", "CDEO"],3, "CODE", ))
```

```python
# 5번
def solution(member_age, member_age_len, transportation):
    transportation_way = { # 어른, 어린이
        "Bus" : [40000, 15000],
        "Ship" :[30000, 13000],
        "Airplane" : [70000, 45000],
        "discount" : [1-0.1, 1-0.2] 
    }
    total = 0
    for i in member_age:
        total += transportation_way[transportation][0 if i>=20 else 1] * (transportation_way["discount"][0 if i>=20 else 1] if member_age_len >= 10 else 1)
    return int(total)

print(solution([13, 33, 45, 11, 20] , 5, "Bus"))
print(solution([25, 11, 27, 56, 7, 19, 52, 31, 77, 8], 10, "Ship"))
```
