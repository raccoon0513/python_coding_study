```python
# Q1. 소용돌이 수
# -*- coding: utf-8 -*-
# UTF-8 encoding when using korean
def solution(n):
    arr = [[0]*n for _ in range(n)]
    x,y= 0,0
    route_axis = 0 # route_axis % 4
    route = [[1, 0], [0,1],[-1,0], [0,-1]]
    for i in range(1, n**2):
        arr[x][y] = i
        nx, ny= x + route[route_axis%4][0], y + route[route_axis%4][1]
        if not ((0 <= nx < n) and (0 <= ny < n) and (arr[nx][ny]==0)):
            route_axis+=1
            nx, ny= x + route[route_axis%4][0], y + route[route_axis%4][1]
        x, y = nx, ny
    arr[x][y] = (n**2)
    return sum([arr[i][i] for i in range(n)])

n1=3
ret1 = solution(n1)
print(f"solution 함수의 반환 값은 {ret1}입니다")

n2=2
ret1 = solution(n2)
print(f"solution 함수의 반환 값은 {ret1}입니다")
```

```python
# Q2. 체스의 나이트
# -*- coding: utf-8 -*-
# UTF-8 encoding when using korean
def solution(pos):
    ascii_arr = "ABCDEDFGH"
    x = ascii_arr.index(pos[0])
    y = int(pos[1]) - 1
    count = 0
    # 반시계 회전 
    moveset_x = [-1, -2, -2 , -1, 1, 2, 2, 1]
    moveset_y = [2, 1, -1, -2, -2, -1, 1, 2]
    for i in range(8):
        if (0<=x+moveset_x[i]<8) and (0<=y+moveset_y[i]<8) : 
            count += 1 
    return count

pos = "A7"
ret = solution(pos)
try:
    print(f"ret : {ret}")
except Exception as e:
    pass


```

```python
# Q3. 타임머신
# -*- coding: utf-8 -*-
# UTF-8 encoding when using korean
def solution(num):
	temp_num = num
	answer = 0
	num_arr = [] # 역순으로 저장됨
	while(temp_num > 0):
		num_arr.append(temp_num%10)
		temp_num = temp_num//10
	upper_flag = True
	for i in range(len(num_arr)):
		if upper_flag:
			num_arr[i] += 1
			if num_arr[i] == 10:
				num_arr[i] = 1
				upper_flag = True	
			else :
				upper_flag = False
	if upper_flag:
		num_arr.append(1)
	ten_key = 1
	for i in num_arr:
		answer+= i * ten_key
		ten_key*=10
	return answer
num = 9949999
ret = solution(num)
 
print("solution 함수의 반환 값은", ret, "입니다.")
```

```python
# Q5. 회원할인
# -*- coding: utf-8 -*-
# UTF-8 encoding when using korean
# s : 5% g : 10%, v:15%
import random as rand
def solution(price, grade):
    sale = {
        "S":1 - (5*0.01),
        "G":1 - (10*0.01),
        "V":1 - (15*0.01)
    }
    return round(price * sale[grade])

for i in range(10):
    price, grade = rand.randint(100, 100000), rand.choice("SGV")
    print(f"grade : {grade}\tprice : {price}\tresult = {solution(price, grade)}" )
```

```python
# Q6. 셔츠 개수
# -*- coding: utf-8 -*-
# UTF-8 encoding when using korean
import random as rand
def solution(shirt_size):
    
    shirt_list = ["XS", "S", "M", "L", "XL", "XXL"]
    answer = [0] * (len(shirt_list))
    for i in shirt_size:
        answer[shirt_list.index(i)] +=1
    return answer

for _ in range(10):
    rand_num = rand.randint(1, 100)
    rand_arr = []
    for i in range(rand_num):
        rand_arr.append(rand.choice(["XS", "S", "M", "L", "XL", "XXL"]))
    print(rand_arr)
    print(solution(rand_arr))
```

```python
def generate_permutations(arr):
    if len(arr) == 1:
        return [arr]

    result = []
    for i in range(len(arr)):
        pivot = arr[i]
        remaining = arr[:i] + arr[i+1:]
        print(f"arr[i] : {arr[i]}, arr[:i] : {arr[:i]}, arr[i+1:] : {arr[i+1:]}")
        for p in generate_permutations(remaining):
            result.append([pivot] + p)
            
    return result

data = [1, 2, 3]
print(generate_permutations(data))
```
